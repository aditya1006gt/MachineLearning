```python
# --- THE MATH FUNCTIONS ---
print("Absolute value:", abs(-99)) 
print("Quotient and Remainder:", divmod(10, 3)) # Returns (10//3, 10%3)
print("Maximum:", max(1, 5, 10, 2))
print("Minimum:", min(1, 5, 10, 2))
print("Power (2 to the power of 3):", pow(2, 3)) # Same as 2 ** 3
print("Round to 2 decimals:", round(3.14159, 2))
print("Sum of a list:", sum([10, 20, 30]))

# Advanced Math (Creates complex numbers used in engineering)
print("Complex number:", complex(2, 3))
```

```python
# --- THE CONVERSION FUNCTIONS ---
print("To Integer:", int("42"))
print("To Float (Decimal):", float(42))
print("To String (Text):", str(42))
print("To Boolean (True/False):", bool(1)) # 0 is False, anything else is True

# Converting into different types of sequences
my_string = "Adi"
print("To List:", list(my_string))
print("To Tuple (unchangeable list):", tuple(my_string))
print("To Set (removes duplicates):", set([1, 2, 2, 3, 3, 3]))

# hex(), bin(), and oct() convert numbers to different base systems
print("To Binary:", bin(10)) 
print("To Hexadecimal:", hex(255))
```

```python
# --- THE SEQUENCE FUNCTIONS ---
names = ["Adi", "John", "Zack", "Alice"]
scores = [100, 85, 95, 70]

print("Length of list:", len(names))
print("Sorted alphabetically:", sorted(names))

# reversed() flips the list backwards
print("Reversed list:", list(reversed(names)))

# enumerate() gives you the index number AND the item in a loop!
print("Enumerate example:")
for index, name in enumerate(names):
    print(f"   Person {index}: {name}")

# zip() stitches two lists together like a zipper!
print("Zip example:")
for name, score in zip(names, scores):
    print(f"   {name} got a {score}")
```

```python
# --- THE LOGIC FUNCTIONS ---
bool_list = [True, True, False]

# all() checks if EVERY item is True
print("Are ALL true?:", all(bool_list))

# any() checks if AT LEAST ONE item is True
print("Are ANY true?:", any(bool_list))

# filter() keeps only the items that pass a test
def is_even(n):
    return n % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(is_even, numbers))
print("Filtered even numbers:", even_numbers)

# map() applies a function to every item in a list
def double_it(n):
    return n * 2

doubled_numbers = list(map(double_it, numbers))
print("Mapped/Doubled numbers:", doubled_numbers)
```
