```python
# indexing

a="Aditya"
print("Indexing " ,end=" ")
for i in a:
    print(i,end=" ")

# Types of Indexing
# Postive Indexing
print("\nPostive Indexing " ,end=" ")
for i in range(6):
    print(a[i],end=" ")

# Negative Indexing
print("\nNegative Indexing " ,end=" ")
for i in range(-6,0):
    print(a[i],end=" ")
print()
```

```python
# Slicing
c = "Hello World"
print (c)
print (c[0:5])
print (c[2:])
print(c[:4])
print (c[:])
print(c[2:6:2])
print(c[0:8:3])
print(c[-5:-1:2]) # for -ve indexing
print (c[::-1]) # important for reversing a string
```
