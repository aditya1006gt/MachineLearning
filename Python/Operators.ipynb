```python
# Arithmetic Operator

a = 10
b = 3

print("Addition (+):", a + b)
print("Subtraction (-):", a - b)
print("Multiplication (*):", a * b)

# Division always gives a decimal number (float)
print("Division (/):", a / b)

# Floor division gives the whole number, chopping off the decimal
print("Floor Division (//):", a // b)

# Modulo gives the remainder (10 divided by 3 is 9, with 1 left over)
print("Modulo/Remainder (%):", a % b)

# Exponentiation (a to the power of b)
print("Exponentiation (**):", a ** b)
```

```python
# Comparison Operator

x = 10
y = 5

print("Is x equal to y? (==) :", x == y)
print("Is x NOT equal to y? (!=) :", x != y)

print("Is x greater than y? (>) :", x > y)
print("Is x less than y? (<) :", x < y)

print("Is x greater than or equal to y? (>=) :", x >= y)
print("Is x less than or equal to y? (<=) :", x <= y)
```

```python
# Logical Operators

a = True
b = False

print("What is 'a and b'? (and) :", a and b) # Both must be True to get True
print("What is 'a or b'? (or) :", a or b)   # At least one must be True to get True
print("What is the opposite of a? (not) :", not a) # Reverses the result
```

```python
# Assignment Operators

x = 10
print("Initial value of x :", x)

x += 5  # This is the same as writing: x = x + 5
print("After x += 5, x is now :", x)

x -= 3  # This is the same as writing: x = x - 3
print("After x -= 3, x is now :", x)

x *= 2  # This is the same as writing: x = x * 2
print("After x *= 2, x is now :", x)

x /= 4  # This is the same as writing: x = x / 4
print("After x /= 4, x is now :", x)
```

```python
# Membership Operators

my_list = [10, 20, 30, 40, 50]
my_name = "Adi"

print("Is 30 inside my_list? (in) :", 30 in my_list)
print("Is 100 inside my_list? (in) :", 100 in my_list)
print("Is 100 NOT inside my_list? (not in) :", 100 not in my_list)

print("Is the letter 'A' in my_name? (in) :", 'A' in my_name)
```

```python
# Identity Operators

list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1 # list3 is now pointing to the exact same list as list1

# '==' checks if the VALUES are the same
print("Do list1 and list2 have the same values? (==) :", list1 == list2) 

# 'is' checks if they are the exact SAME object in memory
print("Are list1 and list3 the exact same object? (is) :", list1 is list3)
print("Are list1 and list2 the exact same object? (is) :", list1 is list2) # False! They look the same, but are two different lists
print("Are list1 and list2 NOT the same object? (is not) :", list1 is not list2)
```

```python
# Bitwise Operators
# (In binary: 5 is 0101, and 3 is 0011)

p = 5 
q = 3 

print("Bitwise AND (&) :", p & q)  # Sets each bit to 1 if both bits are 1
print("Bitwise OR (|) :", p | q)   # Sets each bit to 1 if one of two bits is 1
print("Bitwise XOR (^) :", p ^ q)  # Sets each bit to 1 if only one of two bits is 1
# Bitwise Shift Operators

x = 10  # In binary, 10 is written as 1010

# -----------------------------------------
# RIGHT SHIFT (>>)
# Shifts the bits to the right. 
# Mathematically: 10 divided by 2^1 (which is 2)
# -----------------------------------------
print("Right Shift (10 >> 1) :", x >> 1)  # The result will be 5


# -----------------------------------------
# LEFT SHIFT (<<)
# Shifts the bits to the left.
# Mathematically: 10 multiplied by 2^1 (which is 2)
# -----------------------------------------
print("Left Shift (10 << 1) :", x << 1)   # The result will be 20


# Bonus example: Shifting by 2
print("Right Shift by 2 (10 >> 2) :", x >> 2) # 10 divided by 4 (floor division)
```
