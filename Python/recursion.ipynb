```python
# find a*b uding recursion

def multi(a,b):
    if b==0:
        return 0
    else :
     return multi(a,b-1)+a

print(multi(2,5))
```

```python
# palindrome using recursion

def palin(stg,i,j):
    if i==j:
        return True
    if stg[i]!=stg[j]:
        return False
    return palin(stg,i+1,j-1)

stg=input("String ka value daalo:")
print("String palindrome hai " if palin(stg,0,len(stg)-1) else "String palindrome nhi hai")
```

```python
def fib(a,b,c,month):
    if c==month: 
        return b; 
    return fib(b,a+b,c+1,month);

month=int(input("The month at which you want to find the number of rabbits"))
print(fib(1,1,0,month))
```
